```python
import zipfile
with zipfile.ZipFile("Ecommerce Counterfeit Products Dataset.zip", "r") as zip_ref:
    zip_ref.extractall("ecommerce_dataset")
```

The dataset contains two CSV files: **counterfeit_products.csv** (5,000 rows, 27 columns) and **_counterfeit_transactions.csv** (3,000 rows, 20 columns). We selected counterfeit_products.csv as it contains the is_counterfeit label along with product- and seller-level features directly relevant to detecting counterfeit listings, while the transactions file lacks any authenticity label.

```python
import os
for root, dirs, filenames in os.walk("ecommerce_dataset"):
    for f in filenames:
        print(os.path.join(root, f))
```

```python
import pandas as pd

products_df = pd.read_csv("ecommerce_dataset/counterfeit_products.csv")
transactions_df = pd.read_csv("ecommerce_dataset/_counterfeit_transactions.csv")

print("PRODUCTS columns:", list(products_df.columns))
print("PRODUCTS shape:", products_df.shape)
print()
print("TRANSACTIONS columns:", list(transactions_df.columns))
print("TRANSACTIONS shape:", transactions_df.shape)
```

```python
print("PRODUCTS full column list:")
for col in products_df.columns:
    print(" -", col)

print()
print("TRANSACTIONS full column list:")
for col in transactions_df.columns:
    print(" -", col)
```
